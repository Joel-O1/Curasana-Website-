```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

np.random.seed(42)
random.seed(42)

log_data = []
start_date = datetime(2026, 1, 1)

# Generate 3 distinct patient profiles to create real density
for i in range(1, 100):
    date_offset = timedelta(days=random.randint(0, 30))
    
    # Cohort 1: Migraine + Fatigue users
    log_data.append({"user_id": f"migraine_user_{i}", "timestamp": start_date + date_offset, "symptom": "Migraine", "severity": 4})
    log_data.append({"user_id": f"migraine_user_{i}", "timestamp": start_date + date_offset, "symptom": "Fatigue", "severity": 3})
    
    # Cohort 2: Nausea + Brain Fog users
    log_data.append({"user_id": f"gastro_user_{i}", "timestamp": start_date + date_offset, "symptom": "Nausea", "severity": 5})
    log_data.append({"user_id": f"gastro_user_{i}", "timestamp": start_date + date_offset, "symptom": "Brain Fog", "severity": 3})
    
    # Cohort 3: Joint Pain + Insomnia users
    log_data.append({"user_id": f"arthritic_user_{i}", "timestamp": start_date + date_offset, "symptom": "Joint Pain", "severity": 4})
    log_data.append({"user_id": f"arthritic_user_{i}", "timestamp": start_date + date_offset, "symptom": "Insomnia", "severity": 4})

# Inject 50 rows of pure chaotic random NOISE (unrelated single symptoms)
for k in range(50):
    log_data.append({
        "user_id": f"noisy_user_{k}",
        "timestamp": start_date + timedelta(days=random.randint(0, 30)),
        "symptom": random.choice(["Migraine", "Fatigue", "Nausea", "Brain Fog", "Joint Pain", "Insomnia"]),
        "severity": random.randint(1, 5)
    })

df = pd.DataFrame(log_data)
print(f"Dataset rebuilt with {len(df)} clinically segmented rows.")
```

```python
# 1. Group by user and date, and create a binary matrix of symptoms
# This converts row-based logs into a wide table of 1s and 0s
matrix_df = pd.crosstab(
    [df['user_id'], df['timestamp']], 
    df['symptom']
).reset_index()

# 2. Let's see what this look like
print("📊 Your One-Hot Encoded Symptom Matrix:")
print(matrix_df.head())
```

```python
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler

# 1. Isolate just the symptom columns (exclude user_id and timestamp)
features = matrix_df.drop(columns=['user_id', 'timestamp'])

# 2. Run DBSCAN with refined parameters
# A distance of 0.1 means rows must be nearly identical to group together
dbscan_final = DBSCAN(eps=0.1, min_samples=5, metric='hamming')
final_clusters = dbscan_final.fit_predict(features)

matrix_df['cluster'] = final_clusters

print(f"Unique clusters found: {set(final_clusters)}")
print(f"\nBreakdown per cluster:\n{matrix_df['cluster'].value_counts()}")
```

```python
# 1. Group the matrix by cluster and calculate the mean of each symptom
profile_df = matrix_df.groupby('cluster')[features.columns].mean()

print("🔍 CLINICAL SYMPTOM PROFILES PER CLUSTER:\n")

# 2. Iterate through each cluster and print out the dominant symptoms (threshold > 50%)
for cluster_id, row in profile_df.iterrows():
    if cluster_id == -1:
        print(f"🔹 Cluster {cluster_id} (Pure Noise/Anomalies):")
        print("   -> Catch-all for random, unmapped symptoms.\n")
        continue
        
    # Get symptoms that appear in more than 50% of the logs for this cluster
    dominant_symptoms = row[row > 0.5].index.tolist()
    dominant_symptoms_str = ", ".join(dominant_symptoms)
    
    print(f"🔹 Cluster {cluster_id} Profile:")
    print(f"   -> Dominant Symptoms: [{dominant_symptoms_str}]")
    print(f"   -> Confidence: {row[dominant_symptoms].values.round(2).tolist()}\n")
```

```python
import pandas as pd
from sklearn.cluster import DBSCAN

def analyze_symptom_cooccurrence(raw_log_df: pd.DataFrame) -> dict:
    """
    Takes a raw DataFrame of health logs and extracts population-wide
    symptom co-occurrence profiles using DBSCAN.
    """
    # 1. Pivot raw data into binary feature matrix
    matrix = pd.crosstab(
        [raw_log_df['user_id'], raw_log_df['timestamp']], 
        raw_log_df['symptom']
    ).reset_index()
    
    features = matrix.drop(columns=['user_id', 'timestamp'], errors='ignore')
    
    # 2. Run your optimized DBSCAN parameters
    dbscan_engine = DBSCAN(eps=0.1, min_samples=5, metric='hamming')
    matrix['cluster'] = dbscan_engine.fit_predict(features)
    
    # 3. Extract profiles
    profile_means = matrix.groupby('cluster')[features.columns].mean()
    
    insights = {}
    for cluster_id, row in profile_means.iterrows():
        # Skip noise in the primary output dictionary
        if cluster_id == -1:
            continue
            
        dominant_symptoms = row[row > 0.5].index.tolist()
        
        # Only save multi-symptom clusters (true co-occurrences)
        if len(dominant_symptoms) > 1:
            insights[f"profile_{cluster_id}"] = {
                "symptoms": dominant_symptoms,
                "occurrence_count": int((matrix['cluster'] == cluster_id).sum())
            }
            
    return insights

# Test your new production function instantly using your mock data
compiled_insights = analyze_symptom_cooccurrence(df)
print("🚀 PRODUCTION ENGINE COMPILED SUCCESSFULLY. SAMPLE OUTPUT:")
print(compiled_insights)
```

```python
from sklearn.ensemble import IsolationForest
import numpy as np

def detect_user_anomalies(raw_log_df: pd.DataFrame, target_user_id: str) -> dict:
    """
    Checks an individual user's baseline. If they have logged enough events,
    uses an Isolation Forest to flag severe or highly unusual health days.
    """
    # 1. Isolate the target user's records
    user_df = raw_log_df[raw_log_df['user_id'] == target_user_id].copy()
    
    # 2. Cold Start Protection: Require at least 10 entries
    if len(user_df) < 10:
        return {
            "status": "cold_start",
            "message": f"User has only logged {len(user_df)} events. Need at least 10 for a baseline.",
            "anomalies": []
        }
        
    # 3. Pivot their data by day to look at daily patterns
    user_matrix = pd.crosstab(user_df['timestamp'], user_df['symptom'], values=user_df['severity'], aggfunc='max').fillna(0)
    
    # 4. Train the Isolation Forest on the user's history
    # contamination=0.1 means we expect roughly 10% of their logs to be flagged as unusual outliers
    iso_forest = IsolationForest(contamination=0.1, random_state=42)
    
    # Predict: 1 = normal day, -1 = anomalous day
    user_matrix['anomaly_score'] = iso_forest.fit_predict(user_matrix)
    
    # 5. Extract the anomalous days
    anomaly_days = user_matrix[user_matrix['anomaly_score'] == -1].index.strftime('%Y-%m-%d').tolist()
    
    return {
        "status": "active",
        "total_logs_analyzed": len(user_df),
        "anomalies_detected_on": anomaly_days
    }

# --- TEST THE COLD START GUARDFRAIL ---
# Pick a random noisy user who has barely logged anything
cold_start_test = detect_user_anomalies(df, "noisy_user_0")
print("🧪 COLD START TEST RESULT:")
print(cold_start_test)
```

```python
# Pick one of the heavy users from our cohort who has multiple logs
active_user_test = detect_user_anomalies(df, "migraine_user_1")

print("🌲 ACTIVE USER ANOMALY DETECTION TEST RESULT:")
import json
print(json.dumps(active_user_test, indent=2))
```

```python
# 1. Create a single user with 60 days of consistent baseline logs
super_user_data = []
base_date = datetime(2026, 1, 1)

# Baseline: 60 days of mild-to-moderate tension headaches and fatigue
for day in range(60):
    log_date = base_date + timedelta(days=day)
    # Most days they just log mild fatigue
    super_user_data.append({"user_id": "super_user_007", "timestamp": log_date, "symptom": "Fatigue", "severity": 2})
    # Every few days they get a mild headache
    if day % 3 == 0:
        super_user_data.append({"user_id": "super_user_007", "timestamp": log_date, "symptom": "Migraine", "severity": 1})

# Anomalies: Inject 3 specific days where their health completely drops off a cliff
# (Severe symptoms they don't normally get)
super_user_data.append({"user_id": "super_user_007", "timestamp": datetime(2026, 1, 15), "symptom": "Joint Pain", "severity": 5})
super_user_data.append({"user_id": "super_user_007", "timestamp": datetime(2026, 2, 10), "symptom": "Nausea", "severity": 5})
super_user_data.append({"user_id": "super_user_007", "timestamp": datetime(2026, 2, 28), "symptom": "Migraine", "severity": 5}) # Normal symptom, abnormal severity

# 2. Append this super user into your main dataframe
super_user_df = pd.DataFrame(super_user_data)
df = pd.concat([df, super_user_df], ignore_index=True)

print(f"✅ Injected 'super_user_007' with {len(super_user_df)} logs into the master dataset.")
```

```python
#--------------GENERATED SUMMARY ---------------------------
def generate_doctor_summary(raw_log_df: pd.DataFrame, target_user_id: str, days_back: int = 60) -> dict:
    #create a summary for a users health history over a spcific timeframe into a clean breakdown

    #isolate the user data and filter the requested time frame
    user_df = raw_log_df[raw_log_df['user_id'] == target_user_id].copy()
    max_date = user_df['timestamp'].max()
    cutoff_date = max_date - pd.Timedelta(days=days_back)
    recent_df = user_df[user_df['timestamp'] >= cutoff_date]

    if len(recent_df) == 0:
        return {"message": "No data found for this user in the specified timeframe."}

    #calculate the core frequencies. how often does each symptom happen
    symptom_counts = recent_df['symptom'].value_counts().to_dict()

    #calculate average severity per symptom
    severity_averages = recent_df.groupby('symptom')['severity'].mean().round(1).to_dict()

    #pull the active anomolies over this period using the previous engine
    anomaly_results = detect_user_anomalies(raw_log_df, target_user_id)
    recent_anomalies = []
    if anomaly_results['status'] == 'active':
        # Only keep anomalies that fall within our doctor window
        recent_anomalies = [d for d in anomaly_results['anomalies_detected_on'] if d >= cutoff_date.strftime('%Y-%m-%d')]

    return {
        "user_id": target_user_id,
        "timeframe_days": days_back,
        "total_events_logged": len(recent_df),
        "symptom_frequencies": symptom_counts,
        "average_severities": severity_averages,
        "clinical_anomalies_detected": recent_anomalies
    }

    # Test the summary compiler on our super user
    doctor_export = generate_doctor_summary(df, "super_user_007", days_back=60)
    print("📋 CLINICAL DOCTOR EXPORT DATA CONTRACT:")
    import json
    print(json.dumps(doctor_export, indent=2))
        
```
